```python
#Pandas:基于Numpy,开源数据分析与操作工具
#两大核心：
# 2.DataFrame:二维数组对象，类似numpy中的矩阵，但功能更丰富。类似数据库的表格
# 1.Series:一维数组对象，类似numpy中的数组，但功能更丰富。类似数据库的表格的一列
import pandas as pd

#构造DataFrame-->构建数据集
df = pd.DataFrame([
    {"name":"张三","Chinese":80,"Math":90,"English":100},
    {"name":"李四","Chinese":90,"Math":80,"English":90},
    {"name":"王五","Chinese":100,"Math":95,"English":80},
    {"name":"赵六","Chinese":80,"Math":90,"English":100},
    {"name":"孙七","Chinese":90,"Math":80,"English":90},
    {"name":"周八","Chinese":100,"Math":100,"English":80},
    {"name":"吴九","Chinese":86,"Math":90,"English":100},
    {"name":"郑十","Chinese":90,"Math":80,"English":90},
    {"name":"王十一","Chinese":100,"Math":98,"English":80}
])
df
#统计计算
print(f"语文最高分：{df['Chinese'].max()},语文最低分：{df['Chinese'].min()},语文平均分：{df['Chinese'].mean():.2f}")
print(f"数学最高分：{df['Math'].max()},数学最低分：{df['Math'].min()},数学平均分：{df['Math'].mean():.2f}")
print(f"英语最高分：{df['English'].max()},英语最低分：{df['English'].min()},英语平均分：{df['English'].mean():.2f}")
```
