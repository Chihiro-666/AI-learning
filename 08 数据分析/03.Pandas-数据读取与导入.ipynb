```python
df3 = pd.read_csv('data/movie_list.csv')

#配置项，展示所有数据
# pd.set_option('display.max_rows', None)

#数据过滤-->df3[条件]
df3[df3['评分']>85]   #筛选评分大于85的列
df3[df3['评分']>85][['电影名称','评分']]   #筛选评分大于85的行
df3[df3['语言'].isin(['英语'])] # 筛选语言等于英语的行
df3[df3['类型'].str.contains('动作')]   #筛选类型包含动作的行
df3[df3['类型'].str.contains('动作')][['电影名称','类型']]   #筛选类型包含动作的行
df3[df3["评分"].between(80,90,inclusive="neither")]   #筛选评分在80-90之间的行left right neither
df3[(df3["评分"]>80) & (df3["时长"]<150)]   #筛选评分大于80，时长小于150的行--->和&  或|

```
