```python
import pandas as pd

#1.读取数据
df = pd.read_csv("data/sales.csv")

#2.数据清洗
#2.1 缺失值处理
df.isnull() #查看缺失值

#2.1.1删除缺失值
df.dropna() #删除缺失值所在行
df.dropna(axis=1) #删除缺失值所在列
df

#2.1.2填充缺失值
df.fillna('--') #填充缺失值
df.ffill()   #用前一个值填充
df.bfill()   #用后一个值填充

#2.2 重复值处理
#2.2.1 查看重复值
df.duplicated()  #查看重复值（所有列全都一样）
df.duplicated(subset=['订单号'])  #查看重复值（指定列）

#2.2.2 删除重复值
df.drop_duplicates()  #删除重复值
df.drop_duplicates(subset=['订单号'],keep='first')  #删除重复值（指定列） first/last/False

#2.3 异常值处理
#2.3.1 查看异常值
df[df['单价']<=0]

#2.3.2 删除异常值
df.drop(df[df['单价']<=0].index)

#2.3.3 修复异常值
df['单价'] = df['单价'].abs()
df

# 2.4 数据格式转换
df['订单日期'] = df['订单日期'].replace('2020-01-01', '2020-01-01')  #单个替换
df['订单日期'] = df['订单日期'].str.replace('/', '-')  #批量替换
df
```
