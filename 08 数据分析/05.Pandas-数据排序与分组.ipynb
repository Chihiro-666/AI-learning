```python
import pandas as pd

#1.读取数据
df = pd.read_csv("data/sales.csv",nrows=20)

#2.数据排序
df.sort_values("订单号")   # 根据订单号升序排序，ascending=True默认升序
df.sort_values("单价",ascending=False)   # 根据单价降序排序
df.sort_values(["单价","订单号"],ascending=[False,True])  # 先根据单价降序排序，后根据订单号升序排序
```

```python
#3.数据分组
df["销售额"] = df["单价"] * df["销售数量"]
df.groupby("产品类别")["订单号"].sum()  # 对"产品类别"列进行分组，对销售额求和
df.groupby("产品类别")["订单号"].count()  # 对"产品类别"列进行分组，对订单号统计数量
df.groupby("产品类别")["销售数量"].max()  # 对"产品类别"列进行分组，并对销售额求最大值
df.groupby("产品类别")["销售数量"].min() # 对"产品类别"列进行分组，并对销售额求最小值
df.groupby("产品类别")["销售额"].mean()  # 对"产品类别"列进行分组，并对销售额求平均值
df.groupby("产品类别")["单价"].agg(["mean","max","min"])  # 对"产品类别"列进行分组，并对单价求平均值、最大值、最小值
df.groupby("产品类别").agg({'销售数量':"sum",'销售额':"sum",'单价':"mean"})
df.groupby("产品类别").agg({'销售数量':"sum",'销售额':"sum",'单价':"mean"}).to_dict()  # 将结果转换为字典
```
