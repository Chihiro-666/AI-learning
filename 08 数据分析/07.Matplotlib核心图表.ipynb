```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt
from matplotlib.lines import lineStyles

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

#创建子图
#figure: 画布对象； axes: 子图对象
fig,axes = plt.subplots(nrows=1,ncols=2,figsize=(20,6),dpi=100) #nrows行 ncols列 index子图编号

#图一：柱状图
countries = ['中国','美国','日本','印度','俄罗斯','法国','英国','德国','意大利','西班牙']
values = [118, 130, 106, 84, 117, 77, 80, 50, 65, 60]

axes1: Axes = axes[0]
axes1.bar(countries,values,width=0.6,color='y')
axes1.set_title('世界石油储备',fontsize=20)  #标题
axes1.set_xlabel('国家',fontsize=15)  #x轴标签
axes1.set_ylabel('油量（亿桶）',fontsize=15)  #y轴标签
axes1.grid(linestyle='--', alpha=0.4)  #添加网格,透明度

#图二：饼状图
countries = ['中国','美国','日本','印度','俄罗斯','法国','英国','德国','意大利','西班牙']
#国家对应人数
values = [13.89, 3.30, 12.6, 10.8, 8.9, 7.5, 10.6, 16.9, 6.5, 9.01]

axes2: Axes = axes[1]
axes2.pie(values, labels=countries, autopct='%1.2f%%')
axes2.set_title('2019年世界人口前10国家', fontsize=15)  # 设置标题
axes2.legend(loc='lower center',ncol=5, bbox_to_anchor=(0.5,-0.1))  # 设置图例:loc:位置; ncol:列数 bbox_to_anchor(0.5:水平居中, -0.1:垂直居下)

#保存图片
plt.savefig('data/01.png')

plt.show()
```
